```python
import random

# =====================================================
# Section A — Warm-Up (Comments, Input/Output, Variables)
# =====================================================

# ---------- Problem 1: Comment It ----------
# This line prints a welcome message to the screen.
print("Welcome to today's English lesson!")

# ---------- Problem 2: Introduce Yourself ----------
print("Jeonghyu Park")
print("English Education")
print("Social Study")

# ---------- Problem 3: Ask and Answer ----------
fav_word = input("What is your favorite English word? ")
print("Your favorite word is:", fav_word)

# ---------- Problem 4: Two Inputs, One Sentence ----------
english_word = input("Enter an English word: ")
meaning = input("Enter its meaning: ")
print(english_word, "means", meaning)

# ---------- Problem 5: Vocabulary Card ----------
word = "friend"
part_of_speech = "noun"
meaning = "친구"
print(f"{word} ({part_of_speech}): {meaning}")

# ---------- Problem 6: Total Score ----------
quiz1_score = 85
quiz2_score = 92
print("Total:", quiz1_score + quiz2_score)

# ---------- Problem 7: Average Score ----------
print("Average:", (quiz1_score + quiz2_score) / 2)

# ---------- Problem 8: Word Length Difference ----------
word1 = "apple"
word2 = "banana"
print("Length difference:", len(word1) - len(word2))


# =====================================================
# Section B — Functions & Modules
# =====================================================

# ---------- Problem 9: Double It ----------
def double_score(score):
    return score * 2

print(double_score(5))
print(double_score(21))
print(double_score(100))

# ---------- Problem 10: Percentage Function ----------
def to_percentage(correct, total):
    return correct / total * 100

print(to_percentage(8, 10))
print(to_percentage(3, 4))

# ---------- Problem 11: Longer Word Function ----------
def is_longer(word1, word2):
    return len(word1) > len(word2)

print(is_longer("banana", "apple"))   # True
print(is_longer("cat", "elephant"))   # False

# ---------- Problem 12: Random Greeting ----------
greetings = ["Hello!", "Hi there!", "Good morning!"]
print(random.choice(greetings))

# ---------- Problem 13: Random Number Practice ----------
mystery_number = random.randint(1, 100)
print(mystery_number)

# ---------- Problem 14: Flashcard Maker Function ----------
def make_flashcard(word, meaning):
    return f"{word} -> {meaning}"

print(make_flashcard("apple", "사과"))
print(make_flashcard("book", "책"))


# =====================================================
# Section C — Combining Everything
# =====================================================

# ---------- Problem 15: Pass or Fail (using if) ----------
correct = int(input("How many correct? "))
total = int(input("How many total? "))
percentage = correct / total * 100

if percentage >= 60:
    print("Pass")
else:
    print("Fail")

# ---------- Problem 16: Grade Message Function ----------
def pass_or_fail(percentage):
    if percentage >= 60:
        return "Pass"
    else:
        return "Fail"

print(pass_or_fail(85))
print(pass_or_fail(60))
print(pass_or_fail(42.5))

# ---------- Problem 17: Repeat With a Loop (using for) ----------
words = ["apple", "book", "friend"]

for w in words:
    print(w, "-", len(w), "letters")

# ---------- Problem 18: Random Word Picker + Function ----------
def pick_random_word(word_list):
    return random.choice(word_list)

five_words = ["apple", "book", "friend", "school", "happy"]
print(pick_random_word(five_words))

# ---------- Problem 19: Mini Vocabulary Quiz (one question) ----------
question_word = "apple"
correct_meaning = "사과"

def check_answer(user_answer, correct_answer):
    return user_answer.strip().lower() == correct_answer.strip().lower()

user_answer = input(f"What does '{question_word}' mean? ")

if check_answer(user_answer, correct_meaning):
    print("Correct!")
else:
    print("Try again!")

# ---------- Problem 20: Capstone: 3-Question Vocabulary Quiz ----------
word_list = ["apple", "book", "friend", "school", "happy"]
meaning_list = ["사과", "책", "친구", "학교", "행복한"]

# check_answer()는 Problem 19에서 정의한 함수를 그대로 사용

score = 0
total_questions = 3

for i in range(total_questions):
    index = random.randint(0, len(word_list) - 1)
    quiz_word = word_list[index]
    quiz_meaning = meaning_list[index]

    answer = input(f"Q{i + 1}. What does '{quiz_word}' mean? ")

    if check_answer(answer, quiz_meaning):
        print("Correct!")
        score += 1
    else:
        print(f"Wrong. The answer is {quiz_meaning}.")

print(f"Final score: {score}/{total_questions}")
```
